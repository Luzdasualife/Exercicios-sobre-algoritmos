### Questão 01

As organizações CSM resolveram dar um aumento de salário aos seus colaboradores e lhe contrataram para desenvolver o programa que calculará os reajustes.  Faça um programa que recebe o salário de um colaborador e o reajuste segundo o seguinte critério, baseado no salário atual;

*   Salários até `R$ 280,00` (incluindo): aumento de 20%
*   Salários entre `R$ 280,00` e `R$700,00`: aumento de 15%
*   Salários entre `R$ 700,00` e `R$ 1500,00`: aumento de 10%
*   Salários de `R$ 1500,00` em diante: aumento de 5%

Após o aumento ser realizado; informe na tela;

* O salário antes do reajuste;
* O percentual de aumento aplicado;
* O valor do aumento;
* O novo salário, após o aumento.  

```python
sal = float(input('Informe o salário do colaborador: R$ '))

if sal <= 280.00:
  aumento = (sal * 20/100)
  nsal = aumento + sal
  perc = '20%'

elif 280 < sal < 700:
  aumento = (sal * 15/100)
  nsal = aumento + sal
  perc = '15%'

elif 700 < sal < 1500:
  aumento = (sal * 10/100)
  nsal = aumento + sal
  perc = '10%'

elif sal > 1500:
  aumento = (sal * 5/100)
  nsal = aumento + sal
  perc = '5%'

print('O salário antes do reajuste era: R$',sal)
print('O percentual de aumento aplicado foi:',perc)
print('O valor do aumento é: R$',aumento)
print('O novo salário, após o aumento é: R$',nsal)
```

## Questão 02

Faça um programa que receba a idade de 15 pessoas, e que calcule e mostre a quantidade de pessoas em cada faixa etária e a percentagem de pessoas na primeira e na última faixa etária, de acordo com o seguinte critério:

* Até 15 anos
* De 16 a 30 anos
* De 31 a 45 anos
* De 46 a 60 anos
* Acima de 61 anos

```python
idade = []
A15 = 0
A30 = 0
A45 = 0
A60 = 0
A61 = 0

for i in range(15):
  idades = int(input('informe a idade: '))
  idade.append(idades)
  if idades <= 15:
    A15 += 1
  elif idades > 15 and idades <= 30:
    A30 += 1
  elif idades > 30 and idades <= 45:
    A45 += 1
  elif idades > 45 and idades <= 60:
    A60 += 1
  elif idades > 60:
    A61 += 1

total = len(idade)

if A15 != 0:
  prim = (A15*100)/total
else:
  prim = 'nenhuma'

if A61 != 0:
  ult = (A61*100)/total
else:
  ult = 'nenhuma'

print('A quantidade de pessoas em cada faixa etária é, respectivamente: \nAté 15 anos:',A15,'\nDe 16 a 30 anos:',A30,'\nDe 31 a 45 anos:',A45,'\nDe 46 a 60 anos:',A60,'\nE acima de 61 anos:',A61)

print('a percentagem de pessoas na primeira e na última faixa etária é:\nDe até 15 anos:',prim,'\nAcima de 61 anos:',ult)
```

## Questão 03
Faça um programa que receba um conjunto de valores inteiros e positivos, com o intervalo formado pelo número imprima na tela todos os números primos existentes

```python
num1 = int(input('informe o primeiro numero inteiro e positivo: '))
num2 = int(input('informe o segundo numero inteiro e positivo: '))
primos = []
for i in range(num1, num2 + 1):
 if i > 2 and not any(i % k == 0 for k in range(2, i)):
  primos.append(i)

print(primos)
```

## Questão 04

Para cada casa consultada foi fornecido o número do canal (4, 5, 7, 12) e o número de pessoas que estavam assistindo aquele canal. Se a televisão estivesse desligada, nada era anotado, ou seja, essa casa não entrava na pesquisa. Faça um programa que:  

Leia um número indeterminado de dados (número do canal e o número de pessoas que estavam assistindo);
Calcule e mostre a percentagem de audiência de cada canal.
Para encerrar a entrada de dados digite o número do canal ZERO.

```python
pes4 = 0
pes5 = 0
pes7 = 0
pes12 = 0
total = 0
cal = int(input('Informe o número do canal 4,5,7,12 ou o numero 0 para encerrar a coleta:'))
if cal == 4 or cal == 5 or cal == 7 or cal ==12:
  while cal != 0:
    pes = int(input('Informe o número de pessoas assistindo:'))
    total += pes

    if cal == 4:
      pes4 += pes
    elif cal == 5:
      pes5 += pes
    elif cal == 7:
      pes7 += pes
    elif cal == 12:
      pes12 += pes
    else:
      print('Canal invalido, digite novamente')

    cal = int(input('Informe o número do canal 4,5,7,12 ou o numero 0 para encerrar a coleta:'))
else:
  print('Começe com um canal valido.')

perc4 = (pes4*100)/total
perc5 = (pes5*100)/total
perc7 = (pes7*100)/total
perc12 = (pes12*100)/total
print('A percentagem de audiência de cada canal é:\nDo canal 4:',perc4,'%\nDo canal 5:',perc5,'%\nDo canal 7:',perc7,'%\nDo canal 12:',perc12,'%')




```

## Questão 05
Faça um programa que apresente o menu de opções a seguir:  

Menu de opções:  

1. Média aritmética
2. Média ponderada
3. Sair

Digite a opção desejada:

* Na opção 1: receber duas notas, calcular e mostrar a média aritmética.
* Na opção 2: receber três notas e seus respectivos pesos, calcular e mostrar a média ponderada.
* Na opção 3: sair do programa.  

```python
print('Menu de opções:\n1.Média aritmética\n2.Média ponderada\n3.Sair\n')
op = int(input('Digite a opção desejada: '))
if op == 1:
  n = []
  n.append(float(input('Informe a primeira nota: ')))
  n.append(float(input('Informe a segunda nota: ')))
  media = sum(n)/len(n)
  print('A média aritimética é:',media,'\nAté mais!')
elif op == 2:
  mp = []
  sp = []
  for i in range(3):
    nota = float(input('informe a nota: '))
    p = float(input('informe o peso da nota: '))
    mp.append(nota*p)
    sp.append(p)
  mediap = sum(mp)/sum(sp)
  print('A média ponderada é:',mediap,'\nAté mais!')
elif op == 3:
  print('Até mais!')
else:
  print('Até mais!')


```
